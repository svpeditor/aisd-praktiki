Клики и независимые множества (Брон-Кербош)

```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['A', 'B', 'C', 'F'],
    'E': ['C', 'F', 'G'],
    'F': ['D', 'E', 'G', 'H'],
    'G': ['E', 'F', 'H'],
    'H': ['F', 'G']
}


def bron_kerbosch(R, P, X, graph, cliques):
    if not P and not X:
        cliques.append(R)
        return
    for v in list(P):
        bron_kerbosch(R | {v}, P & set(graph[v]), X & set(graph[v]),
                      graph, cliques)
        P.remove(v)
        X.add(v)


def complement(graph):
    # ребро есть в дополнении, если его нет в исходном графе
    comp = {v: [] for v in graph}
    for u in graph:
        for v in graph:
            if u != v and v not in graph[u]:
                comp[u].append(v)
    return comp
```

```python
cliques = []
bron_kerbosch(set(), set(graph), set(), graph, cliques)
cliques = sorted([sorted(c) for c in cliques], key=lambda c: (-len(c), c))
print('Максимальные клики:')
for c in cliques:
    print('  ', c, len(c))
print('Наибольшая клика:', cliques[0], ', размер', len(cliques[0]))

comp = complement(graph)
comp_edges = sorted({tuple(sorted((u, v))) for u in comp for v in comp[u]})
print()
print('Дополнение графа, ребер', len(comp_edges), ':')
print(' ', ' '.join(u + v for u, v in comp_edges))
```

```python
# независимые множества G = клики дополнения
ind = []
bron_kerbosch(set(), set(comp), set(), comp, ind)
ind = sorted([sorted(s) for s in ind], key=lambda s: (-len(s), s))
print('Максимальные независимые множества:')
for s in ind:
    print('  ', s, len(s))

print()
print('Нижняя оценка: chi(G) >=', len(cliques[0]))
```
