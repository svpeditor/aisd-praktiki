Гамильтонов цикл: поиск с возвратом, теорема Дирака

```python
import math

graph = {
    'A': ['B', 'C', 'F'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['C', 'D', 'F'],
    'F': ['A', 'D', 'E']
}
calls = 0  # сколько раз вызывался backtrack


def hamiltonian_cycle(graph):
    global calls
    vertices = list(graph.keys())
    n = len(vertices)

    def backtrack(path):
        global calls
        calls += 1
        if len(path) == n:
            if path[0] in graph[path[-1]]:
                return path + [path[0]]
            return None
        current = path[-1]
        for nb in graph[current]:
            if nb not in path:   # вершина уже в пути - отсекаем
                result = backtrack(path + [nb])
                if result:
                    return result
        return None

    for start in vertices:
        result = backtrack([start])
        if result:
            return result
    return None
```

```python
cycle = hamiltonian_cycle(graph)
print('Гамильтонов цикл:', ' -> '.join(cycle) if cycle else 'не найден')
print('Вызовов backtrack:', calls)

# теорема Дирака: deg(v) >= n/2 для всех v
n = len(graph)
print()
print('n =', n, ', n/2 =', n / 2)
for v in graph:
    ok = 'ok' if len(graph[v]) >= n / 2 else 'меньше n/2'
    print(' ', v, 'deg =', len(graph[v]), ok)
print('Условие Дирака выполнено:', all(len(graph[v]) >= n / 2 for v in graph))

print()
print('Вариантов обхода (n-1)!/2 =', math.factorial(n - 1) // 2)
print('Для n = 20: (n-1)!/2 =', math.factorial(19) // 2)
```
