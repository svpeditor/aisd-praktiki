Проверка эйлеровости и алгоритм Флери

```python
from collections import deque

graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['A', 'B', 'D', 'E', 'F'],
    'D': ['A', 'C', 'F'],
    'E': ['B', 'C', 'F', 'G'],
    'F': ['C', 'D', 'E', 'G'],
    'G': ['E', 'F']
}


def fleury(graph):
    g = {v: list(nb) for v, nb in graph.items()}
    odd = [v for v in g if len(g[v]) % 2 != 0]
    if len(odd) == 0:
        start = next(iter(g))   # эйлеров цикл, начинаем с любой вершины
    elif len(odd) == 2:
        start = odd[0]          # эйлеров путь, начинаем с нечетной вершины
    else:
        return None

    def is_bridge(u, v):
        # удаляем ребро и смотрим, достижима ли v из u
        g[u].remove(v)
        g[v].remove(u)
        visited = {u}
        q = deque([u])
        while q:
            x = q.popleft()
            for y in g[x]:
                if y not in visited:
                    visited.add(y)
                    q.append(y)
        g[u].append(v)
        g[v].append(u)
        return v not in visited

    route = [start]
    current = start
    while any(g.values()):
        neighbors = g[current]
        next_v = None
        for nb in neighbors:
            if not is_bridge(current, nb):   # сначала не мосты
                next_v = nb
                break
        if next_v is None and neighbors:
            next_v = neighbors[0]
        if next_v is None:
            break
        g[current].remove(next_v)
        g[next_v].remove(current)
        current = next_v
        route.append(current)
    return route


def check_route(graph, route):
    # каждое ребро должно быть пройдено ровно один раз
    used = []
    for i in range(len(route) - 1):
        used.append(frozenset((route[i], route[i + 1])))
    all_edges = {frozenset((u, v)) for u in graph for v in graph[u]}
    return len(used) == len(all_edges) and set(used) == all_edges
```

```python
print('Степени вершин:')
for v in graph:
    print(' ', v, len(graph[v]))
odd = [v for v in graph if len(graph[v]) % 2 != 0]
print('Вершины нечетной степени:', odd, '- всего', len(odd))
if len(odd) == 0:
    print('Граф эйлеров')
elif len(odd) == 2:
    print('Граф полуэйлеров')
else:
    print('Граф не эйлеров и не полуэйлеров')

route = fleury(graph)
print('Результат алгоритма Флери:', route)
if route:
    print('Проверка маршрута:', check_route(graph, route))
```
