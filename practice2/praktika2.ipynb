# Практическая работа 2. Алгоритм Дейкстры
Проверка ручных решений

```python
import networkx as nx
import matplotlib.pyplot as plt
```

## Задача 1

```python
G1 = nx.DiGraph()
G1.add_weighted_edges_from([(1, 2, 1), (2, 3, 4), (3, 1, 2), (1, 4, 7), (3, 4, 13),
                            (4, 5, 1), (5, 3, 12), (5, 6, 5), (6, 3, 2)])
length, path = nx.single_source_dijkstra(G1, source=1, target=6)
print('Длина кратчайшего пути из 1 в 6:', length)
print('Путь:', path)
print('Все метки:', nx.single_source_dijkstra_path_length(G1, 1))
```

## Задача 2

```python
G2 = nx.Graph()
G2.add_weighted_edges_from([(1, 2, 3), (1, 4, 7), (1, 5, 8), (2, 3, 1), (2, 4, 4),
                            (3, 4, 2), (3, 6, 3), (4, 6, 4), (4, 5, 3), (5, 6, 9)])
length, path = nx.single_source_dijkstra(G2, source=1, target=6)
print('Длина кратчайшего пути из 1 в 6:', length)
print('Путь:', path)
```

## Задача 3

```python
G3 = nx.DiGraph()
G3.add_weighted_edges_from([('S', 'A', 5), ('S', 'B', 10), ('A', 'C', 3),
                            ('A', 'T', 12), ('B', 'C', 2), ('C', 'T', 4)])
length, path = nx.single_source_dijkstra(G3, source='S', target='T')
print('Время доставки, мин:', length)
print('Маршрут:', path)

pos = {'S': (0, 1), 'A': (1, 2), 'B': (1, 0), 'C': (2, 0.6), 'T': (3, 1.4)}
nx.draw(G3, pos, with_labels=True, node_color='#cdeeaa', node_size=1000)
nx.draw_networkx_edge_labels(G3, pos, edge_labels=nx.get_edge_attributes(G3, 'weight'))
plt.show()
```
