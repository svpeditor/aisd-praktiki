Компоненты связности и проверка двудольности

```python
from collections import deque

graph = {
    'A': ['B', 'C'], 'B': ['A', 'C'], 'C': ['A', 'B', 'D'], 'D': ['C'],
    'E': ['F'], 'F': ['E', 'G'], 'G': ['F'],
    'H': [],
    'I': ['J', 'L'], 'J': ['I', 'K'], 'K': ['J', 'L'], 'L': ['I', 'K']
}


# одна компонента = один запуск DFS (через стек)
def find_components(graph):
    visited = set()
    components = []
    for vertex in graph:
        if vertex not in visited:
            component = set()
            stack = [vertex]
            while stack:
                v = stack.pop()
                if v not in visited:
                    visited.add(v)
                    component.add(v)
                    for nb in graph[v]:
                        if nb not in visited:
                            stack.append(nb)
            components.append(component)
    return components


# BFS-раскраска в два цвета
def is_bipartite(graph, start):
    color = {start: 0}
    queue = deque([start])
    while queue:
        v = queue.popleft()
        for nb in graph[v]:
            if nb not in color:
                color[nb] = 1 - color[v]
                queue.append(nb)
            elif color[nb] == color[v]:
                # соседние вершины одного цвета
                return False, (v, nb)
    return True, color
```

```python
components = find_components(graph)
print('Компонент связности:', len(components))
for comp in components:
    print(comp)

print()
for comp in components:
    start = sorted(comp)[0]
    ok, res = is_bipartite(graph, start)
    if ok:
        print(sorted(comp), 'двудольная, раскраска:', res)
    else:
        print(sorted(comp), 'не двудольная, конфликт на ребре', res)
```
