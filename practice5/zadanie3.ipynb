Топологическая сортировка и кратчайший путь от A до H

```python
from collections import deque

# ориентированный ациклический граф
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['E', 'F'],
    'D': ['G'],
    'E': ['G', 'H'],
    'F': ['G'],
    'G': ['H'],
    'H': []
}


def topological_sort(graph):
    visited = set()
    result = []

    def dfs(v):
        visited.add(v)
        for nb in graph.get(v, []):
            if nb not in visited:
                dfs(nb)
        # вершину добавляем после всех её потомков
        result.append(v)

    for v in graph:
        if v not in visited:
            dfs(v)
    return result[::-1]


# BFS с запоминанием предшественников
def bfs_path(graph, start, end):
    visited = {start}
    parent = {start: None}
    queue = deque([start])
    while queue:
        v = queue.popleft()
        if v == end:
            break
        for nb in graph[v]:
            if nb not in visited:
                visited.add(nb)
                parent[nb] = v
                queue.append(nb)
    # восстанавливаем путь с конца
    path = []
    v = end
    while v is not None:
        path.append(v)
        v = parent[v]
    return path[::-1]
```

```python
print('Топологический порядок:', topological_sort(graph))

path = bfs_path(graph, 'A', 'H')
print('Кратчайший путь A -> H:', ' -> '.join(path))
print('Длина пути (рёбер):', len(path) - 1)
```
