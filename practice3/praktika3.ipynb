# Практическая работа 3. Максимальный поток
Проверка ручных решений (Форд-Фалкерсон)

```python
import networkx as nx
```

```python
def solve(edges, s, t):
    G = nx.DiGraph()
    for u, v, c in edges:
        G.add_edge(u, v, capacity=c)
    value, flow = nx.maximum_flow(G, s, t)
    cut, (S, T) = nx.minimum_cut(G, s, t)
    print('Максимальный поток:', value)
    print('Минимальный разрез:', cut, S, T)
```

## Задача 1

```python
solve([(1, 2, 3), (1, 4, 5), (4, 2, 4), (2, 3, 5), (4, 5, 2), (3, 5, 6), (3, 6, 5), (5, 6, 7)], 1, 6)
```
