Сеть железных дорог между городами (алгоритм Краскала)

```python
cities = ['SPb', 'Moscow', 'NNovgorod', 'Kazan', 'Ufa', 'Ekaterinburg',
          'Tyumen', 'Samara', 'Volgograd', 'Rostov', 'Krasnodar']
# дороги (город, город, км)
roads = [
    ('SPb', 'Moscow', 650),
    ('SPb', 'NNovgorod', 1200),
    ('Moscow', 'NNovgorod', 400),
    ('Moscow', 'Kazan', 800),
    ('Moscow', 'Rostov', 1100),
    ('NNovgorod', 'Kazan', 400),
    ('Kazan', 'Ufa', 500),
    ('Kazan', 'Samara', 450),
    ('Ufa', 'Samara', 450),
    ('Ufa', 'Ekaterinburg', 550),
    ('Ekaterinburg', 'Tyumen', 350),
    ('Samara', 'Volgograd', 850),
    ('Samara', 'Rostov', 900),
    ('Volgograd', 'Rostov', 400),
    ('Rostov', 'Krasnodar', 250)
]


class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        # сжатие путей
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)
        if rx == ry:
            return False
        # объединение по рангу
        if self.rank[rx] < self.rank[ry]:
            rx, ry = ry, rx
        self.parent[ry] = rx
        if self.rank[rx] == self.rank[ry]:
            self.rank[rx] += 1
        return True


def kruskal(cities, roads):
    idx = {c: i for i, c in enumerate(cities)}
    dsu = DSU(len(cities))
    mst = []
    total = 0
    for u, v, w in sorted(roads, key=lambda r: r[2]):
        if dsu.union(idx[u], idx[v]):
            mst.append((u, v, w))
            total += w
    return mst, total
```

```python
mst, total = kruskal(cities, roads)
print('Выбранные пути:')
for u, v, w in mst:
    print('  ', u, '-', v, w)
print('Ребер:', len(mst), ', общая длина:', total, 'км')

# степени городов в MST
deg = {c: 0 for c in cities}
for u, v, w in mst:
    deg[u] += 1
    deg[v] += 1
items = ['%s %d' % (c, deg[c]) for c in cities]
print('Степени в MST:', ', '.join(items[:6]))
print('              ', ', '.join(items[6:]))
m = max(deg.values())
central = [c for c in cities if deg[c] == m]
print('Центральные города (степень %d):' % m, central)
```
