Флойд-Уоршелл: кратчайшие пути между всеми парами вершин

```python
INF = float('inf')
# вершины 1..5 -> индексы 0..4
# вес дуги 2 -> 4 на рисунке закрыт подписью -2 дуги 4 -> 2, взял 5
graph = [
    [0,   5,   10,  INF, INF],
    [INF, 0,   3,   5,   INF],
    [INF, INF, 0,   2,   7],
    [INF, -2,  INF, 0,   4],
    [12,  INF, INF, INF, 0]
]


def floyd_warshall(graph):
    n = len(graph)
    dist = [row[:] for row in graph]
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
    return dist


def print_matrix(m):
    print('   ' + ''.join('%5d' % (j + 1) for j in range(len(m))))
    for i in range(len(m)):
        row = ''
        for x in m[i]:
            row += '%5s' % ('inf' if x == INF else x)
        print('%3d' % (i + 1) + row)
```

```python
print('Матрица смежности:')
print_matrix(graph)

dist = floyd_warshall(graph)
print()
print('Матрица кратчайших расстояний:')
print_matrix(dist)

# отрицательный цикл есть, если на диагонали получилось меньше 0
neg = [i + 1 for i in range(len(dist)) if dist[i][i] < 0]
print()
if neg:
    print('Есть отрицательный цикл, в него входят вершины:', neg)
else:
    print('Отрицательных циклов нет, на диагонали все нули')
```
