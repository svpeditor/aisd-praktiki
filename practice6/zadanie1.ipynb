Беллман-Форд: кратчайшие пути от S, восстановление пути до T

```python
# вершины: S=0, A=1, B=2, C=3, D=4, T=5
names = ['S', 'A', 'B', 'C', 'D', 'T']
edges = [
    (0, 1, 4),   # S -> A
    (0, 2, 3),   # S -> B
    (1, 2, -2),  # A -> B
    (1, 3, 7),   # A -> C
    (2, 3, 5),   # B -> C
    (2, 4, 8),   # B -> D
    (3, 4, -1),  # C -> D
    (3, 5, 6),   # C -> T
    (4, 5, 3)    # D -> T
]
n = 6


def bellman_ford(edges, n, start):
    INF = float('inf')
    dist = [INF] * n
    prev = [None] * n  # предшественники для восстановления пути
    dist[start] = 0
    for _ in range(n - 1):
        updated = False
        for u, v, w in edges:
            if dist[u] != INF and dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                prev[v] = u
                updated = True
        if not updated:
            break
    # проверка отрицательного цикла
    for u, v, w in edges:
        if dist[u] != INF and dist[u] + w < dist[v]:
            print('Есть отрицательный цикл!')
            return None, None
    return dist, prev


def get_path(prev, end):
    # идем от конца по предшественникам
    path = []
    v = end
    while v is not None:
        path.append(v)
        v = prev[v]
    return path[::-1]
```

```python
dist, prev = bellman_ford(edges, n, 0)
if dist is not None:
    print('Отрицательного цикла нет')
    print('Массив расстояний:', dist)
    for i in range(n):
        print('  S ->', names[i], ':', dist[i])
    path = get_path(prev, 5)
    print('Кратчайший путь S -> T:', ' -> '.join(names[v] for v in path))
    print('Длина пути:', dist[5])
```
